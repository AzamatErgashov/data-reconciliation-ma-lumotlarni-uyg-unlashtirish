# sodda misollar 

```python
A = pd.DataFrame({"Transport": ['poyezd','avtomobil','kema'],
     "Tezligi": ['130','100','60']})
```

```python
A
```

```python
B = pd.DataFrame({"Transport": ['poyezd','avtomobil','tranvay'],
     "Narxi": ['13000','10000','6000']})
```

# data setlar sodda bulganligi uchun bu tasavvur qilishimiz uchun qulay buladi

# endi datalarni merge qilib kuramiz


```python
merge_df = pd.merge(A, B)#inter join ichki qushilish
```

```python
merge_df
```

```python
merge_df = pd.merge(A, B, how='outer')# outer join tashqi qushilish
```

```python
merge_df
```

# left maqsadi chap data yani A da uxshash bulmagan datalar ham olinadi Bda esa faqat uxshashlar olinadi

```python
merge_left = pd.merge(A, B, how='left')# left join chapga birlashish
```

```python
merge_left
```

```python
merge_right = pd.merge(A, B, how='right')# right join ongdan birlashish
```

```python
merge_right
```

```python
A = pd.DataFrame({"Transport_turlari": ['poyezd','avtomobil','kema'],
     "Tezligi": ['130','100','60']})
```

```python
B = pd.DataFrame({"Transport": ['poyezd','avtomobil','tranvay'],
     "Narxi": ['13000','10000','6000']})
```

# agar bizda ustun nomlari bir xil bulmasa hato hisoblaydi buni tugirlash uchun qaysi ustun bilan boglashni aytish kerak

```python
meg = pd.merge(A,B, left_on="Transport_turlari", right_on='Transport')
meg
```
